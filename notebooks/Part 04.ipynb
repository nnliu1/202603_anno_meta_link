### 4.2 Mapping Free-text to OEO Thesaurus

```python
raw_keywords = ["PV System", "Wind Farm", "Coal Station"]
# Mapping to OEO labels
oeo_mapping = {
    "PV System": "oeo:photovoltaic power plant",
    "Wind Farm": "oeo:wind power plant",
    "Coal Station": "oeo:coal power plant"
}

mapped_data = [oeo_mapping.get(k, "Unmapped_Term") for k in raw_keywords]
print(f"Mapped to OEO: {mapped_data}")
```

### 4.3 Semantic Annotation of Tabular Data 



```python
import pandas as pd

# Raw energy production data
df = pd.DataFrame({
    'timestamp': ['2026-03-15T09:00:00Z'],
    'gen_mw': [450.5],
    'source': ['Solar']
})

# Semantic Metadata Mapping
semantic_context = {
    'gen_mw': {
        'property': 'oeo:has_magnitude',
        'type': 'oeo:electrical power generation',
        'unit': 'oeo:megawatt'
    }
}

print(f"Semantic Annotation for 'gen_mw': {semantic_context['gen_mw']}")
```
