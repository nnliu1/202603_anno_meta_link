## 4. Mapping & Semantic Annotating of Data



### 4.1 Why Do We Need Annotate Tabular Data with Ontology Terms?

#### Core Problem
Tabular data (CSV, Excel, Database tables) is everywhere in energy research, without ontology annotations, machines cannot understand what the data represents. 

#### Key Benefits

1. **Integration into a (virtual) knowledge graph**
    Annotated tables can be mapped to RDF triples via R2RML. And through a **Virtual Knowledge Graph**, you don't even have to move your data anywhere. Your data stays in your local database, but it becomes queryable as part of a larger knowledge graph. 

2. **Scenario comparison across different dataset**
    Example: OEP scenario bundles

3. **Schema matching for data integration**
    When merging datasets from different sources, semantic annotations enable automated schema matching rather than manual mapping

4. **Reproducibility & Provenance**
    Annotations provide a machine-readable description of what each column means, enabling reproducible analysis pipelines

### 4.2 Mapping Free-text to OEO Terms

Energy datasets often contain informal or domain-specific text labels. These need to be mapped to standardised vocabulary to enable interoperability

- "solar", "PV", "Photovoltaics", "solar power plant", these should all resolve to the same ontology class 
- OEO provides a controlled vocabulary


```python
# Raw keywords as they might appear in a dataset (informal labels)
raw_keywords = ["PV System", "Coal Station", "Nuclear Plant"]

# Manual mapping to OEO preferred labels with their IRIs
# OEO IRIs follow the pattern: http://openenergy-platform.org/ontology/oeo/OEO_XXXXXXXX
oeo_mapping = {
    "PV System":      {"label": "photovoltaic power plant",       "iri": "OEO_00000324"},
    "Coal Station":   {"label": "hard coal power plant",          "iri": "OEO_00000205"},
    "Nuclear Plant":  {"label": "nuclear power plant",            "iri": "OEO_00000303"},
}

mapped_data = [oeo_mapping.get(k, "Unmapped_Term") for k in raw_keywords]
print(f"Mapped to OEO: {mapped_data}")
```

### 4.3 Semantic Annotation of Tabular Data
