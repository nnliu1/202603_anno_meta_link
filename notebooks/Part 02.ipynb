## Data Governance (LinkML)

We use [LinkML](https://linkml.io/) to define and validate the sensor payloads. This ensures the data entering the Neo4j graph and the plotter remains consistent.

- **Schema File:** `fenecon_mea.yaml`
- **To validate data:** `uv run linkml-validate -s fenecon_mea.yaml data.json`
- **To update Python classes:** `uv run gen-python fenecon_mea.yaml > datamodel.py`
