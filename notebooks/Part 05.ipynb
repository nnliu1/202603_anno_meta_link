### 2. Build Metadata for Your Data
##### **Hands on Task 2** (10 min)

Convert your observations into structured metadata

```python
import pandas as pd

df = pd.read_csv("iai_active_power_pv_202201.csv")
df.head()
```

```python
## Create Dataset Description
dataset_description = {
    "file_name": "",
    "file_type": "",
    "data_structure": "",
    "columns": list(df.columns),
    "num_rows": len(df),
    "note": ""
}

dataset_description
```

```python
# Fill Metadata 
zenodo_metadata_draft = {
    "title": "",
    "creators": [],
    "description": "",
    "publication_date": "",
    "resource_type": "",
    "license": "",
    "doi_status": ""
}

oep_metadata_draft = {
    "name": "",
    "title": "",
    "description": "",
    "keywords": [],
    "subject": [],
    "publicationDate": "",
    "spatial": {},
    "licenses": [],
    "metaMetadata": {}
}

zenodo_metadata_draft, oep_metadata_draft
```

```python
## Combine Dataset and Metadata
draft_record = {
    "dataset_description": dataset_description,
    "zenodo_metadata": zenodo_metadata_draft,
    "oep_metadata": oep_metadata_draft
}

draft_record
```

### 4. Publish Draft

prepare final upload-ready metadata

#### **Hands on Task 3** (15 min)

1) Define version information 

```python
version_info = {
    "version": "",
    "release_date": "",
    "version_status": "",
    "changes": []
}

version_info
```

2) Create version-aware file names

```python
import os

original_file = dataset_description["file_name"]
base_name, ext = os.path.splitext(original_file)

versioned_dataset_file = f"{base_name}_{version_info['version']}{ext}"
versioned_metadata_file = f"{base_name}_{version_info['version']}_metadata.json"

print("Versioned dataset file:", versioned_dataset_file)
print("Versioned metadata file:", versioned_metadata_file)
```

3) Update dataset description

```python
dataset_description_versioned = dataset_description.copy()

dataset_description_versioned.update({
    "version": version_info["version"],
    "release_date": version_info["release_date"],
    "versioned_file_name": versioned_dataset_file
})

dataset_description_versioned
```

4) Update metadata draft

```python
zenodo_metadata_versioned = zenodo_metadata_draft.copy()

zenodo_metadata_versioned.update({
    "version": version_info["version"],
    "publication_date": version_info["release_date"],
    "notes": ""
})


oep_metadata_versioned = oep_metadata_draft.copy()

oep_metadata_versioned["metaMetadata"]["metadataVersion"] = "OEMetadata-2.0"
oep_metadata_versioned["publicationDate"] = version_info["release_date"]
oep_metadata_versioned["version"] = version_info["version"]

zenodo_metadata_versioned, oep_metadata_versioned
```

5) Create release note

```python
release_note = f"""
Release: {version_info['version']}
Date: {version_info['release_date']}
Status: {version_info['version_status']}

Changes in this release:
- {"; ".join(version_info["changes"])}
"""

print(release_note)
```

6) Build the publication-ready draft

```python
publication_ready_draft = {
    "dataset_file": versioned_dataset_file,
    "metadata_file": versioned_metadata_file,
    "version": version_info["version"],
    "release_note": release_note.strip(),
    "dataset_description": dataset_description_versioned,
    "zenodo_metadata": zenodo_metadata_versioned,
    "oep_metadata": oep_metadata_versioned
}

publication_ready_draft
```

7) Export the metadata draft to JSON

```python
import json

with open(versioned_metadata_file, "w", encoding="utf-8") as f:
    json.dump(publication_ready_draft, f, indent=2, ensure_ascii=False)

print(f"Metadata draft saved as: {versioned_metadata_file}")
```
